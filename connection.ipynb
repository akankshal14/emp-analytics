```python
import pandas as pd
from sqlalchemy import create_engine, text

# --- Credentials (adjust pass/port if needed) ---
DB_USER = "root"
DB_PASS = "I_amsk77"  # Replace with your MySQL password
DB_HOST = "localhost"
DB_PORT = "3306"
DB_NAME = "mini_project"   # Using exact lowercase

# Step 1: Connect to server generally
server_url = f"mysql+pymysql://{DB_USER}:{DB_PASS}@{DB_HOST}:{DB_PORT}"
server_engine = create_engine(server_url)

# Step 2: Ensure database exists
with server_engine.connect() as conn:
    conn.execute(text(f"CREATE DATABASE IF NOT EXISTS {DB_NAME};"))
    conn.commit()
    print(f"Verified Database '{DB_NAME}' exists!")

# Step 3: Connect directly to mini_project
db_url = f"mysql+pymysql://{DB_USER}:{DB_PASS}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
db_engine = create_engine(db_url)

# Step 4: Import 100,000 rows
print("Reading CSV file...")
df = pd.read_csv("IBM_HR_Unified_100k.csv")

print("Importing to MySQL table 'stg_employees' (this takes ~15-20 seconds)...")
df.to_sql("stg_employees", con=db_engine, if_exists="replace", index=False, chunksize=10000)

print(f"Success! 100,000 records imported into database '{DB_NAME}'.")
```
